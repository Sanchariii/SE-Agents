```python
SYSTEM_PROMPT = """You are the Planning Agent of a training-deck creation system. You
turn a user's instruction into a layout-aware DeckPlan by calling your tools — you never
write plan JSON yourself, the tools do that.

A single user turn may bundle SEVERAL distinct instructions at once (e.g. "update slide
3's numbers, remove slide 5, and mark this confidential"). Handle every one of them by
calling the right tool for each, in whatever order makes sense, before giving your final
answer. Don't stop after the first instruction if more remain.

Typical approach:
1. Call read_current_plan first to see what already exists.
2. Decide per instruction:
   - A brand-new deck, or a change to the deck's overall scope/structure -> create_deck_plan.
   - A targeted change to specific existing slide(s) -> revise_deck_slides with those
     slide numbers.
   - A request to drop slide(s) -> remove_deck_slides.
   - An explicit classification (public/internal/confidential/restricted/co-branded)
     mentioned anywhere in the message -> set_deck_classification.
   - A change to which LANGUAGES the deck is produced in ("also in French", "generate
     it in all languages", "English only") -> set_deck_languages. The slides are not
     rewritten for this; they are re-rendered and translated.
    - you can use tools update enricnch slides 
3. If create_deck_plan comes back asking a clarifying question, relay that question to
   the user as your final answer instead of guessing at missing information.

Never invent target slide numbers that aren't in the current plan. End with a concise,
plain-language summary of what you did (or the clarifying question), written for the user
— not a tool-call log."""

_TOOL_INTENT = {
    "create_deck_plan": "plan",
    "add_deck_slides": "add",
    "revise_deck_slides": "edit_or_enrich",
    "set_slide_layout": "relayout",
    "remove_deck_slides": "remove",
    "set_deck_classification": "classification",
    "set_deck_languages": "language",
}


def _agent(session_id: str) -> ToolCallingAgent:
    tools =build_plan_tools(session_id)
    return ToolCallingAgent(name="planner", system_prompt=SYSTEM_PROMPT, tools=tools,
                            tool_intent=_TOOL_INTENT, session_id=session_id)


def _snapshot(session_id: str) -> DeckPlan | None:
    """The stored plan as a value object, so a later mutation cannot alter it."""
    plan_dict, _ = session_store.get_plan(session_id)
    return DeckPlan(**plan_dict) if plan_dict else None
```

```python
feedback =  []

def _deck_context(plan: DeckPlan, position: int) -> DeckContext:
    """Deck-level context for the slide at 1-based `position` — neighbours only, never
    the whole deck, or the per-slide token saving is given straight back."""
    titles = [s.title for s in plan.slides]
    return DeckContext(
        title=plan.title, audience=plan.audience, objective=plan.objective,
        tone=plan.tone, position=position, total=len(plan.slides),
        previous_title=titles[position - 2] if position > 1 else "",
        next_title=titles[position] if position < len(titles) else "")
```

```python
def build_content (task: SlideTask):
    planned = plan_by_number.get(task.slide_number)
    if planned is None:
        return None

    if feedback and task.op != SlideAgentConstants.OP_KEEP:
        task = task.model_copy(update={"feedback": feedback})
    try:
        print(f"runnig the  ppt agent for SLIDE :  {task.slide_number} .. \n")
        return run_ppt_agent(
            sid, task, planned, classification,
            _deck_context(plan, task.slide_number),
            previous=previous.get(task.slide_number),
            chunks=list(chunks_by_number.get(task.slide_number) or []),
            )
    except Exception as exc:  # noqa: BLE001
        # One slide must never take the deck down with it, least of all when the
        # slides are running concurrently.
        print(exc)
        stored = previous.get(task.slide_number)
        return SlideResult(
            slide_number=task.slide_number,
            slide_draft=stored or fallback_slide(planned, catalog),
            chunks=list(chunks_by_number.get(task.slide_number) or []),
            notes=f"slide build failed: {exc}")
```

```python
slides, contents, gaps = [], [], []
for position, planned in enumerate(plan.slides, start=1):
    result = results.get(planned.slide_number)
    source = result.slide_draft if result else old_slides.get(planned.slide_number)
    if source is None:
        print("collect: slide %d has neither a fresh nor a stored draft" % planned.slide_number)
        continue
    # Renumber contiguously so an insertion or removal leaves no gap.
    slide = source.model_copy(deep=True)
    slide.slide_number = position
    slides.append(slide)

    contents.append(SlideContent(
        slide_number=position, title=planned.title,
        chunks=list(result.chunks if result
                    else old_chunks.get(planned.slide_number, [])),
        coverage=result.coverage if result else "good",
        note=result.notes if result else ""))
    if result and result.notes:
        gaps.append(f"Slide {position} ('{planned.title}'): {result.notes}")
```

```python

_ROLE_LAYOUT_KINDS = ("cover", "agenda", "thanks")
def _deck_level_checks(slides: list, classification: str) -> list[str]:
    """Cross-slide faults a per-slide critic cannot see. Deterministic — no LLM."""
    if not slides:
        return []
    notes = []
    seen: dict[str, int] = {}
    for slide in slides:
        key = slide.title.strip().lower()
        if key in seen:
            notes.append(f"Slides {seen[key]} and {slide.slide_number} share the title "
                         f"'{slide.title}'.")
        seen[key] = slide.slide_number

    run_start, run_layout = slides[0].slide_number, slides[0].layout
    for slide in slides[1:] + [None]:
        if slide is not None and slide.layout == run_layout:
            continue
        end = slide.slide_number if slide else slides[-1].slide_number + 1
        if end - run_start >= 3 and run_layout:
            notes.append(f"Layout '{run_layout}' repeats on {end - run_start} "
                         f"consecutive slides from {run_start}.")
        if slide is not None:
            run_start, run_layout = slide.slide_number, slide.layout

    notes += _skeleton_checks(slides, classification)
    return notes


def _skeleton_checks(slides: list, classification: str) -> list[str]:
    """Confirm the published deck has the shape the planner was told to produce.

    Args:
        slides: The assembled SlideDrafts, ordered by slide number.
        classification: Which company template the deck renders on.

    Returns:
        One note per missing, misplaced or wrongly-laid-out role slide.
    """
    kinds = [slide.kind for slide in slides]
    notes = []
    if kinds[0] != "cover":
        notes.append(f"Deck opens on a '{kinds[0]}' slide instead of a cover.")
    if "agenda" not in kinds:
        notes.append("Deck has no agenda slide.")
    if kinds[-1] != "thanks":
        notes.append(f"Deck ends on a '{kinds[-1]}' slide instead of a thank-you.")

    # A role slide sitting on a content layout means its layout_hint never resolved.
    catalog = catalog_mod.for_classification(classification)
    for slide in slides:
        if slide.kind in _ROLE_LAYOUT_KINDS and catalog.resolve(slide.layout).role != slide.kind:
            notes.append(f"Slide {slide.slide_number} is a {slide.kind} slide but "
                         f"renders on layout '{slide.layout}'.")
    return notes


def _render_faults(draft: DeckDraft, audience: str, classification: str) -> list[str]:
    """Render the deck for real and report only the objective faults.

    Deterministic: the report text is scanned here, never sent to a model. Replaces the
    render-report half of the retired deck critic.

    Args:
        draft: The assembled deck, about to be published.
        audience: Deck audience, needed to render the cover.
        classification: Which company template to render on.

    Returns:
        One note per overflowing placeholder, visible file path, off-brand font or
        template mismatch. Empty when the deck is clean or the audit could not run.
    """
    faults = []
    # The policy symbol lives in the master, so the wrong template silently ships the
    # wrong classification marking on every slide.
    wanted = registry.normalize(classification)
    if wanted not in Path(registry.path_for(wanted)).name.lower():
        faults.append(f"Deck rendered on a template that is not '{wanted}'.")

    report = render_report(draft, audience, classification)
    print(report)
    if not report:
        return faults

    current = ""
    for line in report.splitlines():
        if line.startswith("Slide "):
            current = line.split("—")[0].strip()
            continue
        if "OVERFLOWS" in line:
            faults.append(f"{current}: text overflows its placeholder.")
        elif "RAW FILE PATH" in line:
            faults.append(f"{current}: a file path is showing as slide copy.")
        elif "OFF-BRAND FONT" in line:
            faults.append(f"{current}: text is set in a font the brand template does not use.")
        elif "OFF-SLIDE" in line:
            faults.append(f"{current}: a drawn shape falls outside the slide.")
        elif "OVERLAP" in line:
            faults.append(f"{current}: drawn shapes cover each other.")
    return faults
```

```python
"""One-off: rebuild the user's section slide and report what actually rendered.

Run from repo root:
    python -m scripts.manual_testing._analyze_section_slide
"""

import json
import tempfile
from pathlib import Path

from pptx import Presentation
from pptx.util import Emu

from app.core.models.deck_model import DeckDraft
from app.genai.ppt.builder import build_pptx
from app.genai.ppt.template import catalog as catalog_mod, registry

DRAFT = {
    "template": "internal",
    "title": "SMT PPM KPIs: Measuring, Monitoring & Improving SMT Line Performance",
    "subtitle": "Equip participants with a solid understanding of PPM as a defect metric, the 4M KPIV framework, SPC tools, reflow and inspection processes, and how to use AOI feedback for continuous improvement.",
    "takeaways": [],
    "slides": [
        {
            "slide_number": 3,
            "kind": "content",
            "layout": "1_Quote or Chapter / Section ",
            "title": "Section 1: PPM — The Language of SMT Quality",
            "blocks": [
                {"slot": "body1",
                 "text": "How defect rates are defined, measured, and driven down on the SMT line",
                 "bullets": [], "image_ref": ""}
            ],
            "shapes": [
                {"kind": "rounded_rectangle", "left": 60, "top": 380, "width": 203, "height": 460,
                 "text": "PPM Metric", "fill_color": "003E7E", "line_color": "003E7E", "text_color": "FFFFFF"},
                {"kind": "rounded_rectangle", "left": 285, "top": 380, "width": 203, "height": 460,
                 "text": "4M Framework", "fill_color": "CD163F", "line_color": "CD163F", "text_color": "FFFFFF"},
                {"kind": "rounded_rectangle", "left": 510, "top": 380, "width": 203, "height": 460,
                 "text": "SPC Tools", "fill_color": "F58024", "line_color": "F58024", "text_color": "000000"},
                {"kind": "rounded_rectangle", "left": 735, "top": 380, "width": 203, "height": 460,
                 "text": "SPI & AOI", "fill_color": "77C14C", "line_color": "77C14C", "text_color": "000000"},
            ],
            "tables": [],
            "notes": "Section 1 introduces PPM…",
            "bullets": [], "cards": [], "steps": [],
            "stat_value": "", "stat_label": "", "image_path": "",
        }
    ],
}


def main() -> None:
    draft = DeckDraft(**DRAFT)
    classification = registry.normalize(draft.template)
    catalog = catalog_mod.for_classification(classification)

    # What the catalog thinks the requested layout is.
    requested = draft.slides[0].layout
    spec = catalog.resolve(requested)
    print(f"Requested layout: {requested!r}")
    print(f"Resolved layout : {spec.name!r} (role={spec.role}, purpose={spec.purpose})")
    print("Text slots on this layout:")
    for s in spec.text_slots:
        print(f"   - {s.slot:10s} kind={s.kind:8s} {s.width_in:.2f}x{s.height_in:.2f}in "
              f"~{s.capacity} chars")
    print(f"Has 'body1' slot? {spec.slot('body1') is not None}")
    print(f"Title slot: {spec.title.slot if spec.title else None}")
    print()

    with tempfile.TemporaryDirectory() as tmp:
        out = Path(tmp) / "one.pptx"
        build_pptx(draft, "process engineers", out, Path(tmp), classification)
        prs = Presentation(str(out))
        sw, sh = Emu(prs.slide_width).inches, Emu(prs.slide_height).inches
        print(f"Slide size: {sw:.2f} x {sh:.2f} in\n")

        slide = prs.slides[0]
        print(f"Rendered layout: {slide.slide_layout.name!r}")
        print("Shapes on rendered slide:")
        for shape in slide.shapes:
            kind = shape.shape_type
            ph = "PH" if shape.is_placeholder else "  "
            text = ""
            if shape.has_text_frame:
                text = shape.text_frame.text.strip().replace("\n", " / ")[:60]
            l = Emu(shape.left or 0).inches
            t = Emu(shape.top or 0).inches
            w = Emu(shape.width or 0).inches
            h = Emu(shape.height or 0).inches
            print(f"  [{ph}] {str(kind):22s} @ {l:5.2f},{t:5.2f} {w:5.2f}x{h:5.2f}  {text!r}")


if __name__ == "__main__":
    main()
```

```python
from app.genai.ppt.template import catalog as catalog_mod, loader, registry
from app.core.models.deck_model import ContentBlock
from app.genai.ppt.template import filler


def render_content(prs, catalog, slide_draft, resolve_image) -> None:
    spec = catalog.resolve(slide_draft.layout)
    blocks = []
    for block in slide_draft.blocks:
        slot = spec.slot(block.slot)
        if slot is None:
            continue
        if slot.kind == "picture":
            local = resolve_image(block.image_ref) if block.image_ref else None
            if local:
                blocks.append(ContentBlock(slot=block.slot, image_ref=local))
            continue
        blocks.append(block)
    filler.add_slide(
        prs,
        spec,
        title=slide_draft.title,
        blocks=blocks,
        shapes=slide_draft.shapes,
        tables=slide_draft.tables,
        notes=slide_draft.notes,
    )
```
